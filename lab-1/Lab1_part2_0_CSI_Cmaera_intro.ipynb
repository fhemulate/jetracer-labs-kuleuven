```python
from jetcam.csi_camera import CSICamera
from IPython.display import display, Image
import cv2

camera = CSICamera(
    width=640,
    height=360,
    capture_width=1280,
    capture_height=720,
    capture_fps=30
)
```

```python
import ipywidgets as widgets
import traitlets # Traitlets is a Python library that allows two objects to share values and update automatically when something changes.
camera.running = True

image_widget = widgets.Image(format='jpeg', width=640, height=360)

def to_jpeg(frame):
    return cv2.imencode('.jpg', cv2.cvtColor(frame, cv2.COLOR_RGB2BGR))[1].tobytes()

camera_link = traitlets.dlink(
    (camera, 'value'),
    (image_widget, 'value'),
    transform=to_jpeg
)

display(image_widget)
```

```python
camera.cap.release()
```

```python
camera_link.unlink()
```

```python
from jetcam.csi_camera import CSICamera
from PIL import Image
from IPython.display import display

camera = CSICamera(
    width=224,
    height=224,
    capture_width=1640,
    capture_height=1232,
    capture_fps=30
)

frame = camera.read()

print("Saved image size:", frame.shape)
Image.fromarray(frame).save("test_1640x1232.png")
display(Image.fromarray(frame))

camera.cap.release()
```
