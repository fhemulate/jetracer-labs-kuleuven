## 2. Start the car controller

`NvidiaRacecar` has two motion commands:

| Command | Value | Meaning |
|---|---|---|
| `car.steering` | `0.0` | Center position |
| `car.steering` | Negative / positive | Turn in opposite directions |
| `car.throttle` | `0.0` | Stop |
| `car.throttle` | Small positive value | Move forward (after ESC start) |

The values have no unit. They are not angles or speeds in m/s.

```python
import json
from pathlib import Path
import ipywidgets as widgets
from IPython.display import display
from jetracer.nvidia_racecar import NvidiaRacecar

CONFIG_FILE = Path('jetracer_calibration.json')

car = NvidiaRacecar()
car.throttle = 0.0
car.steering = 0.0

print('Car controller is ready. Throttle is 0.')
```

## 3. Read the starting settings

If the JSON file exists, the notebook reads it. Otherwise, it uses the settings below. The values are **starting values**, not universal settings for every car.

For this exercise, change mainly `steering_offset`, `steering_gain`, and `throttle_gain`.

```python
settings = {
    'car_id': 'jetracer_1',
    'pwm_frequency': 60,
    'throttle_gain': 0.8,
    'start_throttle': 0.12,
    'min_running_throttle': 0.11,
    'steering_gain': -0.9,
    'steering_offset': 0.07,
    'steering_calibrated': False,
    'steering_physical_min': -0.7,
    'steering_physical_max': 0.7,
    'steering_min': -0.65,
    'steering_max': 0.65,
    'speed_control_mode': 'throttle',
    'stop': 0.0,
    'start': 0.12,
    'slow': 0.11,
    'normal': 0.13,
    'joystick_throttle_gain': 0.3
}

if CONFIG_FILE.exists():
    with CONFIG_FILE.open('r') as f:
        settings = json.load(f)

car.steering_gain = float(settings['steering_gain'])
car.steering_offset = float(settings['steering_offset'])
car.throttle_gain = float(settings['throttle_gain'])
car.steering_motor.throttle = car.steering_offset

print('Car ID:', settings['car_id'])
print('Steering gain:', car.steering_gain)
print('Steering offset:', car.steering_offset)
print('Throttle gain:', car.throttle_gain)
```

### What do the settings mean?

| Setting | Purpose |
|---|---|
| **Steering offset** | Moves the steering center. Adjust it until the wheels point straight when `car.steering = 0`. |
| **Steering gain** | Changes the steering direction and how far the wheels turn for a given command. A negative value reverses the motor signal. |
| **Throttle gain** | Scales the throttle command. A larger gain gives a stronger motor signal for the same command. |
| **Start throttle** | A short command used to start moving from rest. |
| **Minimum running throttle** | A lower command that may keep the car moving after it starts. |
| **Steering limits** | Restrict the commands to the tested range. |
| **PWM frequency** | Frequency used by the motor controller. The Waveshare `ws/pro` driver sets 60 Hz. |
| **Joystick throttle gain** | Scales the joystick throttle input. It does not replace `car.throttle_gain`. |

The driver applies the following rules:

`steering motor signal = car.steering × steering_gain + steering_offset`

`throttle motor target = car.throttle × throttle_gain`

The Waveshare driver also uses a soft-start method for throttle. The car may not react immediately to each change.

## 4. Calibrate the steering

Keep the wheels above the table. Use the controls below.

1. Set **Steering command** to `0.00`.
2. Change **Steering offset** until the front wheels point straight.
3. Set **Steering command** to `-0.30`, then `+0.30`.
4. If the directions are wrong, change the sign of **Steering gain**.
5. Return **Steering command** to `0.00`. Check the center again.

The steering command is limited to **-0.65 to +0.65**. Do not force the wheels against their mechanical stops.

```python
steering_gain = widgets.FloatSlider(
    value=float(settings['steering_gain']), min=-1.0, max=1.0,
    step=0.05, description='Steering gain', continuous_update=False,
    style={'description_width': 'initial'}
)
steering_offset = widgets.FloatSlider(
    value=float(settings['steering_offset']), min=-0.15, max=0.15,
    step=0.01, description='Steering offset', continuous_update=False,
    style={'description_width': 'initial'}
)
steering_command = widgets.FloatSlider(
    value=0.0, min=float(settings['steering_min']),
    max=float(settings['steering_max']), step=0.05,
    description='Steering command', continuous_update=False,
    style={'description_width': 'initial'}
)
center_button = widgets.Button(description='Center wheels')

def update_steering(change=None):
    car.steering_gain = steering_gain.value
    car.steering_offset = steering_offset.value
    car.steering = steering_command.value
    # Update the motor when only gain or offset changes.
    car.steering_motor.throttle = (
        car.steering * car.steering_gain + car.steering_offset
    )

for slider in (steering_gain, steering_offset, steering_command):
    slider.observe(update_steering, names='value')

center_button.on_click(lambda button: setattr(steering_command, 'value', 0.0))
update_steering()

display(widgets.VBox([
    steering_gain, steering_offset, steering_command, center_button
]))
```

## 5. Try the joystick (steering only)

Connect a gamepad to the **computer that runs the web browser**. Press a button on the gamepad. Its state must appear in the widget.

This step uses **joystick steering only**. It does not enable throttle.

```python
controller = widgets.Controller(index=0)
display(controller)
```

Move the left stick and check that the axis values change. Then run the next cell.

The sign below is `-1.0` because this JetRacer setup previously turned right when the left stick was moved left. If your gamepad already has the correct direction, use `1.0`.

```python
assert len(controller.axes) > 3, 'Connect the joystick before you continue.'

JOYSTICK_STEERING_SIGN = -1.0  # Use 1.0 if your joystick direction is correct.
use_joystick = widgets.Checkbox(value=False, description='Use joystick for steering')

def joystick_to_steering(change):
    if use_joystick.value:
        value = JOYSTICK_STEERING_SIGN * controller.axes[0].value
        value = max(steering_command.min, min(steering_command.max, value))
        steering_command.value = round(value, 2)

controller.axes[0].observe(joystick_to_steering, names='value')
display(use_joystick)
```

## 6. Test low throttle values

Do the first test with the **drive wheels above the table**. Do a floor test only after the instructor checks the car and the area is clear.

- **STOP: 0.00** — Stop the wheels.
- **START: 0.12** — Start moving from rest.
- **SLOW: 0.11** — Try a lower value after START.
- **NORMAL: 0.13** — Slightly stronger command for this introductory test.

The results depend on the car and the battery. These values do not represent a fixed speed.

Use **Allow throttle test** to enable the commands. Press **STOP** when the test ends. If control is lost, turn off the car at its physical switch.

```python
throttle_gain = widgets.FloatSlider(
    value=float(settings['throttle_gain']), min=0.4, max=1.0,
    step=0.05, description='Throttle gain', continuous_update=False,
    style={'description_width': 'initial'}
)
allow_throttle = widgets.Checkbox(value=False, description='Allow throttle test')
speed_buttons = widgets.ToggleButtons(
    options=['STOP', 'START', 'SLOW', 'NORMAL'], value='STOP',
    description='Throttle'
)
stop_button = widgets.Button(description='STOP NOW', button_style='danger')
status = widgets.HTML(value='Throttle is stopped.')

speeds = {
    'STOP': 0.0,
    'START': float(settings['start']),
    'SLOW': float(settings['slow']),
    'NORMAL': float(settings['normal'])
}

def update_speed(change=None):
    if not allow_throttle.value:
        speed_buttons.value = 'STOP'
    speed = speeds[speed_buttons.value]
    speed = min(0.13, max(0.0, speed))  # Limit this lab test.
    car.throttle = speed
    status.value = 'Throttle command: {:.2f}'.format(speed)

def change_gain(change):
    allow_throttle.value = False
    car.throttle = 0.0
    car.throttle_gain = throttle_gain.value

def stop_car(button):
    allow_throttle.value = False
    speed_buttons.value = 'STOP'
    car.throttle = 0.0

allow_throttle.observe(update_speed, names='value')
speed_buttons.observe(update_speed, names='value')
throttle_gain.observe(change_gain, names='value')
stop_button.on_click(stop_car)

car.throttle = 0.0
display(widgets.VBox([
    throttle_gain, allow_throttle, speed_buttons, stop_button, status
]))
```

### Check your results

1. Does the car hold a straight steering position at `0.00`?
2. Does a negative steering command turn the car in the expected direction?
3. Does `START` begin motion? Can `SLOW` keep the car moving?
4. Does `STOP` stop the wheels?

Turn off **Allow throttle test**. Set the steering command to `0.00`.

## 7. Save the settings to JSON

Tick the box only after you check the steering center and direction. Run the next cell to save the values.

Keep the JSON file in the **same folder as both notebooks**. You can open the `.json` file in the JupyterLab text editor to inspect or change the values later.

```python
car_id = widgets.Text(value=settings['car_id'], description='Car ID')
steering_checked = widgets.Checkbox(
    value=False, description='Steering center and direction checked'
)
display(car_id, steering_checked)
```

```python
if not steering_checked.value:
    raise ValueError('Check the steering and tick the box before saving.')

allow_throttle.value = False
car.throttle = 0.0
steering_command.value = 0.0
update_steering()

settings['car_id'] = car_id.value.strip()
settings['steering_gain'] = round(float(steering_gain.value), 2)
settings['steering_offset'] = round(float(steering_offset.value), 2)
settings['throttle_gain'] = round(float(throttle_gain.value), 2)
settings['steering_calibrated'] = True
settings['start_throttle'] = float(settings['start'])
settings['min_running_throttle'] = float(settings['slow'])

if not settings['car_id']:
    raise ValueError('Enter a car ID before saving.')

with CONFIG_FILE.open('w') as f:
    json.dump(settings, f, indent=2)

print('Saved:', CONFIG_FILE.resolve())
```

## 8. Check the saved file

Run the cell below. Check `steering_gain`, `steering_offset`, and `throttle_gain`.

**Result:** The file stores your calibration. The validation notebook will load these values without a new calibration.

```python
with CONFIG_FILE.open('r') as f:
    print(f.read())

car.throttle = 0.0
car.steering = 0.0
print('Done. Turn off the car before you open the validation notebook.')
```
